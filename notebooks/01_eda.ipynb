```python
import pandas as pd

df = pd.read_csv("../data/raw/data.csv")
print(df.shape)
df.head()
```

```python
# Create income brackets for peer grouping
df['Income_Bracket'] = pd.qcut(df['Income'], q=4, labels=['Low', 'Mid-Low', 'Mid-High', 'High'])

# Group by cohort: City_Tier + Income_Bracket
cohort_cols = ['City_Tier', 'Income_Bracket']
category_cols = ['Rent', 'Groceries', 'Transport', 'Eating_Out', 'Entertainment', 'Utilities', 'Healthcare', 'Education', 'Miscellaneous']

cohort_stats = df.groupby(cohort_cols)[category_cols].agg(['mean', 'std'])
cohort_stats.head()
```

```python
def benchmark_user(user_row, df, cohort_stats, category_cols):
    """Compare a single user's spending to their City_Tier + Income_Bracket cohort."""
    tier = user_row['City_Tier']
    bracket = user_row['Income_Bracket']
    cohort = cohort_stats.loc[(tier, bracket)]
    
    results = {}
    for cat in category_cols:
        user_val = user_row[cat]
        cohort_mean = cohort[(cat, 'mean')]
        cohort_std = cohort[(cat, 'std')]
        
        # percentile via z-score approximation
        z = (user_val - cohort_mean) / cohort_std if cohort_std > 0 else 0
        pct_diff = ((user_val - cohort_mean) / cohort_mean) * 100 if cohort_mean > 0 else 0
        
        results[cat] = {
            'user_spend': round(user_val, 2),
            'cohort_avg': round(cohort_mean, 2),
            'pct_diff_from_avg': round(pct_diff, 1),
            'z_score': round(z, 2)
        }
    return results

# Test it on one real user
sample_user = df.iloc[0]
result = benchmark_user(sample_user, df, cohort_stats, category_cols)

for cat, stats in result.items():
    print(f"{cat}: you spend ₹{stats['user_spend']} vs cohort avg ₹{stats['cohort_avg']} ({stats['pct_diff_from_avg']:+.1f}%)")
```
